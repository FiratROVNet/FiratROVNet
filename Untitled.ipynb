```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import ConvexHull
from shapely.geometry import Polygon, Point
from shapely.ops import unary_union

class GüvenlikAlanıYöneticisi:
    def __init__(self, engel_yaricapi=25):
        """
        Args:
            engel_yaricapi: Engelin lastiği ne kadar içeri iteceği (yarıçap).
        """
        self.engel_yaricapi = engel_yaricapi

    def hesapla(self, nokta_bulutu, yasakli_noktalar):
        pts = np.array(nokta_bulutu)
        if len(pts) < 3: return []

        try:
            # 1. Adım: Tüm noktaları kapsayan ana dış çerçeveyi bul
            hull = ConvexHull(pts)
            ana_hull_poly = Polygon(pts[hull.vertices])
            
            # 2. Adım: Engelleri dairesel alanlar olarak tanımla
            engeller = [Point(n).buffer(self.engel_yaricapi) for n in yasakli_noktalar]
            toplam_engel_alani = unary_union(engeller)
            
            # 3. Adım: Çıkarma işlemi (Lastiği içeri bükme)
            nihai_geometri = ana_hull_poly.difference(toplam_engel_alani)

            # Koordinatları geri döndür
            if nihai_geometri.geom_type == 'Polygon':
                return list(nihai_geometri.exterior.coords)
            elif nihai_geometri.geom_type == 'MultiPolygon':
                return [list(poly.exterior.coords) for poly in nihai_geometri.geoms]
            return []
        except Exception as e:
            print(f"Hesaplama hatası: {e}")
            return []

# --- VERİ ÜRETİMİ (Yüzlerce Nokta ve Rastgele Engeller) ---

np.random.seed(42) # Her seferinde aynı rastgelelik için (test amaçlı)

# 1. BÜYÜK NOKTA BULUTU OLUŞTURMA
# -300 ile 300 arasında rastgele 150 nokta üretelim
nokta_sayisi = 150
nokta_bulutu = np.random.uniform(-300, 300, size=(nokta_sayisi, 2))

# 2. RASTGELE ENGEL ÜRETME
# Nokta bulutunun sınırlarını bulalım
x_min, x_max = nokta_bulutu[:, 0].min(), nokta_bulutu[:, 0].max()
y_min, y_max = nokta_bulutu[:, 1].min(), nokta_bulutu[:, 1].max()

# Bu alanın içine serpiştirilmiş 10 adet yasaklı engel noktası
engel_sayisi = 50
rastgele_engeller = np.column_stack((
    np.random.uniform(x_min + 100, x_max - 100, engel_sayisi),
    np.random.uniform(y_min + 100, y_max - 100, engel_sayisi)
))

# 3. SİSTEMİ ÇALIŞTIRMA
yonetici = GüvenlikAlanıYöneticisi(engel_yaricapi=40) # Engellerin itme gücü
guvenli_sinir = yonetici.hesapla(nokta_bulutu, rastgele_engeller)

# --- GÖRSELLEŞTİRME ---
plt.figure(figsize=(12, 10))

# Tüm noktaları çiz (Gri küçük noktalar)
plt.scatter(nokta_bulutu[:, 0], nokta_bulutu[:, 1], c='gray', s=10, alpha=0.5, label='Nokta Bulutu')

# Engelleri çiz (Kırmızı X)
plt.scatter(rastgele_engeller[:, 0], rastgele_engeller[:, 1], 
            marker='x', c='red', s=100, linewidths=2, label='Yasaklı Engeller', zorder=5)

# Dinamik Sınırı Çiz (Kırmızı gerilmiş lastik)
if len(guvenli_sinir) > 0:
    # Çoklu poligon kontrolü
    if isinstance(guvenli_sinir[0], list) and not isinstance(guvenli_sinir[0][0], (int, float)):
        for i, parca in enumerate(guvenli_sinir):
            x, y = zip(*parca)
            plt.plot(x, y, 'r-', linewidth=3, label='Güvenli Sınır' if i==0 else "")
            plt.fill(x, y, color='red', alpha=0.1)
    else:
        x, y = zip(*guvenli_sinir)
        plt.plot(x, y, 'r-', linewidth=3, label='Güvenli Sınır')
        plt.fill(x, y, color='red', alpha=0.1)

plt.title(f"{nokta_sayisi} Nokta ve {engel_sayisi} Rastgele Engel ile Dinamik Alan")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
def point_in_convex_hull(point, hull):
    """
    point: (x, y)
    hull:  CCW sirali convex hull noktalar listesi [(x1,y1), (x2,y2), ...]
    """
    if len(hull) < 3:
        return False

    def cross(o, a, b):
        return (a[0] - o[0]) * (b[1] - o[1]) - (a[1] - o[1]) * (b[0] - o[0])

    sign = None
    for i in range(len(hull)):
        o = hull[i]
        a = hull[(i + 1) % len(hull)]
        cp = cross(o, a, point)

        if cp == 0:
            continue  # kenar uzerinde

        if sign is None:
            sign = cp > 0
        elif (cp > 0) != sign:
            return False

    return True
```

```python
# Optimize Edilmiş HullManager Kodu
"""
Optimize Edilmiş HullManager Sınıfı
3D uzayda güvenli bölgeler oluşturma ve formasyon kontrolü
"""

import numpy as np
import logging
from typing import Optional, List, Tuple, Union, Dict, Any
from dataclasses import dataclass

import numpy.typing as npt
from shapely.geometry import Point, Polygon, MultiPolygon, LineString
from shapely.ops import unary_union, nearest_points
from shapely.prepared import prep
from scipy.spatial.distance import pdist
from scipy.spatial import KDTree

# Scipy Kontrolü - opsiyonel bağımlılık
try:
    from scipy.spatial import ConvexHull
    SCIPY_AVAILABLE = True
except ImportError:
    SCIPY_AVAILABLE = False
    logging.warning("Scipy bulunamadı. ConvexHull özellikleri kısıtlı olacak.")

# Veri Tipleri
Point3D = Tuple[float, float, float]
Point2D = Tuple[float, float]
HullResult = Dict[str, Any]

@dataclass
class HullConfig:
    """Hull yapılandırma parametreleri"""
    default_buffer_radius: float = 12.0
    channel_width: float = 2.0
    safety_factor: float = 2.0  # Güvenlik çarpanı
    tolerance_factor: float = 0.9  # Formasyon toleransı
    min_points_for_hull: int = 3
    z_default: float = 0.0  # 2D işlemler için varsayılan Z değeri

class FakeHull:
    """
    Scipy ConvexHull benzeri arayüz sağlayan wrapper sınıfı.
    Shapely Polygon ile entegre çalışır.
    """
    def __init__(self, points: npt.NDArray[np.float64], polygon: Polygon):
        self.points = points  # Nokta dizisi
        self.polygon = polygon  # Shapely Polygon objesi
        self.vertices = np.arange(len(points))
        self.simplices = self._generate_simplices(len(points))
    
    def _generate_simplices(self, n_points: int) -> npt.NDArray[np.int32]:
        """Kenarları (simplices) oluştur"""
        if n_points < 3:
            return np.array([], dtype=np.int32)
        # Çokgen kenarlarını oluştur
        return np.array([[i, (i + 1) % n_points] 
                        for i in range(n_points)], dtype=np.int32)
    
    def __len__(self) -> int:
        return len(self.points)
    
    def __getitem__(self, index: Union[int, slice]) -> Any:
        return self.points[index]
    
    def __iter__(self):
        return iter(self.points)
    
    def __repr__(self) -> str:
        return f"FakeHull(points={len(self.points)}, area={self.polygon.area:.2f})"

class HullManager:
    """
    3D uzayda güvenli bölgeler oluşturma ve yönetme sınıfı.
    
    Özellikler:
    - Çokgen tabanlı güvenli alan oluşturma
    - Yasaklı bölge entegrasyonu
    - Formasyon nokta validasyonu
    - Performans optimizasyonu
    """
    
    def __init__(
        self, 
        fleet_ref: Optional[Any] = None,
        config: Optional[HullConfig] = None,
        logger: Optional[logging.Logger] = None
    ):
        """
        HullManager başlatıcı
        
        Args:
            fleet_ref: Filo referansı (opsiyonel)
            config: Hull yapılandırması
            logger: Özel logger instance
        """
        self.fleet = fleet_ref
        self.config = config or HullConfig()
        self.logger = logger or logging.getLogger(__name__)
        
        # Performans için cache
        self._prepared_polygons = {}
        
        self.logger.info(f"HullManager başlatıldı. Config: {self.config}")
    
    def create_hull(
        self, 
        base_points: List[Point3D],
        forbidden_points: Optional[List[Point3D]] = None,
        offset: float = 0.0,
        buffer_radius: Optional[float] = None,
        **kwargs
    ) -> HullResult:
        """
        Temel noktalardan güvenli alan (hull) oluşturur.
        Yasaklı noktalardan kanallar açarak geçiş sağlar.
        
        Args:
            base_points: Temel çokgen noktaları (X, Y, Z)
            forbidden_points: Yasaklı bölge noktaları
            offset: Çokgen buffer offset değeri
            buffer_radius: Yasaklı bölge yarıçapı
            
        Returns:
            Dict: {
                'points': numpy array of hull vertices,
                'center': (x, y, z) merkez noktası,
                'hull': FakeHull objesi,
                'area': alan değeri,
                'success': başarı durumu
            }
        """
        # Varsayılan dönüş değeri
        EMPTY_RESULT = {
            'points': None,
            'center': None,
            'hull': None,
            'area': 0.0,
            'success': False
        }
        
        # Giriş validasyonu
        if not self._validate_input_points(base_points):
            self.logger.warning(f"Geçersiz base_points: {len(base_points) if base_points else 0}")
            return EMPTY_RESULT
        
        buffer_radius = buffer_radius or self.config.default_buffer_radius
        
        try:
            # 1. Temel çokgen oluştur
            polygon_2d = self._create_base_polygon(base_points, offset)
            if polygon_2d.is_empty:
                return EMPTY_RESULT
            
            # 2. Yasaklı bölgeleri işle
            if forbidden_points:
                polygon_2d = self._process_forbidden_zones(
                    polygon_2d, forbidden_points, buffer_radius
                )
            
            # 3. Sonuç çokgenini hazırla
            if polygon_2d.is_empty:
                return EMPTY_RESULT
            
            final_polygon = self._get_largest_polygon(polygon_2d)
            
            # 4. Çıktıyı oluştur
            hull_points = np.array(final_polygon.exterior.coords)
            center = self._calculate_center(final_polygon, base_points)
            hull_obj = FakeHull(hull_points, final_polygon)
            
            result = {
                'points': hull_points,
                'center': center,
                'hull': hull_obj,
                'area': float(final_polygon.area),
                'success': True
            }
            
            self.logger.debug(f"Hull oluşturuldu: {len(hull_points)} nokta, area={result['area']:.2f}")
            return result
            
        except Exception as e:
            self.logger.error(f"Hull oluşturma hatası: {e}", exc_info=True)
            return EMPTY_RESULT
    
    def is_formation_valid(
        self, 
        formation_points: List[Point3D], 
        hull: Optional[Union[FakeHull, Polygon]],
        formation_distance: float
    ) -> bool:
        """
        Formasyon noktalarının geçerliliğini kontrol eder.
        
        Kontroller:
        1. Tüm noktalar hull içinde mi?
        2. Noktalar birbirine çok yakın mı?
        
        Args:
            formation_points: Formasyon noktaları listesi
            hull: Güvenli alan referansı
            formation_distance: Minimum formasyon mesafesi
            
        Returns:
            bool: Formasyon geçerliyse True
        """
        # Hızlı kontroller
        if hull is None:
            self.logger.warning("Hull None, formasyon geçersiz")
            return False
        
        if not formation_points:
            self.logger.warning("Boş formasyon noktaları")
            return False
        
        try:
            # 1. Noktaları 2D'ye indirge
            points_2d = np.array([(p[0], p[1]) for p in formation_points])
            
            # 2. Hull içinde mi kontrolü
            if not self._are_points_inside_hull(points_2d, hull):
                self.logger.debug("Bazı noktalar hull dışında")
                return False
            
            # 3. Mesafe kontrolü (çarpışma önleme)
            if len(formation_points) > 1:
                if not self._check_point_distances(
                    points_2d, 
                    formation_distance * self.config.tolerance_factor
                ):
                    self.logger.debug("Noktalar birbirine çok yakın")
                    return False
            
            return True
            
        except Exception as e:
            self.logger.error(f"Formasyon kontrol hatası: {e}", exc_info=True)
            return False  # Hata durumunda güvenli taraf
    
    def is_point_inside(
        self, 
        point: Point3D, 
        hull: Optional[Union[FakeHull, Polygon]]
    ) -> bool:
        """
        Tek bir noktanın hull içinde olup olmadığını kontrol eder.
        
        Args:
            point: Kontrol edilecek nokta (X, Y, Z)
            hull: Güvenli alan referansı
            
        Returns:
            bool: Nokta içerdeyse True
        """
        if hull is None:
            return False
        
        try:
            polygon = self._extract_polygon(hull)
            if polygon is None:
                return False
            
            point_2d = Point(point[0], point[1])
            return polygon.contains(point_2d)
            
        except Exception as e:
            self.logger.error(f"Nokta kontrol hatası: {e}", exc_info=True)
            return False
    
    def calculate_convex_hull(
        self, 
        points: List[Point3D]
    ) -> Optional[npt.NDArray[np.float64]]:
        """
        Scipy kullanarak convex hull hesaplar.
        Fallback: Eğer Scipy yoksa basit bounding polygon.
        
        Args:
            points: Giriş noktaları
            
        Returns:
            Optional[np.ndarray]: Convex hull köşe noktaları
        """
        if not points or len(points) < self.config.min_points_for_hull:
            return None
        
        try:
            # Scipy varsa kullan
            if SCIPY_AVAILABLE:
                points_2d = np.array([(p[0], p[1]) for p in points])
                hull = ConvexHull(points_2d, qhull_options='QJ')
                return points_2d[hull.vertices]
            else:
                # Fallback: Basit bounding box
                return self._simple_bounding_polygon(points)
                
        except Exception as e:
            self.logger.error(f"Convex hull hesaplama hatası: {e}")
            return None
    
    # Yardımcı metodlar
    def _validate_input_points(
        self, 
        points: List[Point3D]
    ) -> bool:
        """Giriş noktalarını valide et"""
        if not points:
            return False
        
        if len(points) < self.config.min_points_for_hull:
            return False
        
        # Tüm noktalar en az 2 koordinata sahip mi?
        return all(len(p) >= 2 for p in points)
    
    def _create_base_polygon(
        self, 
        points: List[Point3D], 
        offset: float
    ) -> Polygon:
        """Temel çokgen oluştur"""
        # Sadece X,Y koordinatlarını al
        points_2d = [(p[0], p[1]) for p in points]
        
        polygon = Polygon(points_2d)
        
        # Validasyon ve buffer işlemleri
        if not polygon.is_valid:
            polygon = polygon.buffer(0)
        
        if offset != 0.0:
            polygon = polygon.buffer(offset)
        
        return polygon
    
    def _process_forbidden_zones(
        self, 
        polygon: Polygon,
        forbidden_points: List[Point3D],
        buffer_radius: float
    ) -> Polygon:
        """Yasaklı bölgeleri işle ve kanallar aç"""
        exclusion_zones = []
        
        for point in forbidden_points:
            if len(point) < 2:
                continue
            
            point_2d = Point(float(point[0]), float(point[1]))
            
            # Sadece yakın noktaları işle
            if polygon.distance(point_2d) < buffer_radius * self.config.safety_factor:
                # Yasaklı alan
                forbidden_zone = point_2d.buffer(buffer_radius)
                
                # Kanala aç
                p1, p2 = nearest_points(forbidden_zone, polygon.exterior)
                channel = LineString([p1, p2]).buffer(self.config.channel_width)
                
                # Birleştir
                exclusion_zones.append(unary_union([forbidden_zone, channel]))
        
        # Tüm yasaklı bölgeleri uygula
        if exclusion_zones:
            combined_exclusion = unary_union(exclusion_zones)
            polygon = polygon.difference(combined_exclusion)
        
        return polygon
    
    def _get_largest_polygon(
        self, 
        geometry: Union[Polygon, MultiPolygon]
    ) -> Polygon:
        """MultiPolygon'dan en büyük poligonu seç"""
        if isinstance(geometry, MultiPolygon):
            return max(geometry.geoms, key=lambda g: g.area)
        return geometry
    
    def _calculate_center(
        self, 
        polygon: Polygon, 
        original_points: List[Point3D]
    ) -> Point3D:
        """Merkez noktasını hesapla"""
        rp = polygon.representative_point()
        
        # Orijinal noktalardan ortalama Z değerini al
        z_values = [p[2] for p in original_points if len(p) > 2]
        avg_z = np.mean(z_values) if z_values else self.config.z_default
        
        return (float(rp.x), float(rp.y), float(avg_z))
    
    def _are_points_inside_hull(
        self, 
        points_2d: npt.NDArray[np.float64],
        hull: Union[FakeHull, Polygon]
    ) -> bool:
        """Tüm noktaların hull içinde olup olmadığını kontrol et"""
        polygon = self._extract_polygon(hull)
        if polygon is None:
            return False
        
        # Prepared polygon for performans
        if id(polygon) not in self._prepared_polygons:
            self._prepared_polygons[id(polygon)] = prep(polygon)
        
        prepared = self._prepared_polygons[id(polygon)]
        
        # Tüm noktaları kontrol et
        for point in points_2d:
            if not prepared.contains(Point(point[0], point[1])):
                return False
        
        return True
    
    def _check_point_distances(
        self, 
        points_2d: npt.NDArray[np.float64],
        min_distance: float
    ) -> bool:
        """Noktalar arası minimum mesafe kontrolü"""
        # Küçük setler için pdist kullan
        if len(points_2d) < 100:
            distances = pdist(points_2d)
            return not np.any(distances < min_distance)
        else:
            # Büyük setler için KDTree
            tree = KDTree(points_2d)
            pairs = tree.query_pairs(min_distance)
            return len(pairs) == 0
    
    def _extract_polygon(
        self, 
        hull: Union[FakeHull, Polygon]
    ) -> Optional[Polygon]:
        """Hull objesinden Polygon çıkar"""
        if isinstance(hull, FakeHull):
            return hull.polygon
        elif isinstance(hull, Polygon):
            return hull
        return None
    
    def _simple_bounding_polygon(
        self, 
        points: List[Point3D]
    ) -> npt.NDArray[np.float64]:
        """Scipy yoksa basit bounding polygon"""
        points_2d = np.array([(p[0], p[1]) for p in points])
        
        # Min-max bounding box
        min_x, min_y = np.min(points_2d, axis=0)
        max_x, max_y = np.max(points_2d, axis=0)
        
        return np.array([
            [min_x, min_y],
            [max_x, min_y],
            [max_x, max_y],
            [min_x, max_y],
            [min_x, min_y]  # Kapat
        ])
    
    def clear_cache(self):
        """Performans cache'ini temizle"""
        self._prepared_polygons.clear()
        self.logger.debug("Hull cache temizlendi")

```

```python
import numpy as np
import math
import logging
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from shapely.geometry import Point, Polygon, MultiPolygon, LineString, MultiPoint, box
from shapely.ops import unary_union, nearest_points
from shapely.prepared import prep
from scipy.spatial.distance import pdist
from typing import List, Tuple, Optional, Union, Dict, Any
from dataclasses import dataclass

# Scipy Kontrolü
try:
    from scipy.spatial import ConvexHull
    SCIPY_AVAILABLE = True
except ImportError:
    SCIPY_AVAILABLE = False

@dataclass
class HullResult:
    """Hull oluşturma sonucunu saklamak için veri sınıfı"""
    points: Optional[np.ndarray] = None
    center: Optional[Tuple[float, float, float]] = None
    hull: Any = None
    polygon: Optional[Polygon] = None
    is_valid: bool = False
    error_message: Optional[str] = None

class SahteHull:
    """
    Hem bir liste gibi davranan hem de Shapely Polygon taşıyan 
    ve len() destekleyen Wrapper sınıfı.
    """
    def __init__(self, points: np.ndarray, polygon_obj: Polygon):
        self.points = points
        self.polygon = polygon_obj
        # Noktaları polygon sınırına göre doğru sırala
        self.vertices = self._order_points_by_polygon(points, polygon_obj)
        # Polygon'ın köşeleri için simplices oluştur
        n = len(points)
        self.simplices = np.array([[i, (i + 1) % n] for i in range(n)])

    def _order_points_by_polygon(self, points: np.ndarray, polygon: Polygon) -> np.ndarray:
        """Noktaları polygon sınırına göre saat yönünde sıralar"""
        if not isinstance(polygon, Polygon):
            return np.arange(len(points))
        
        try:
            # Polygon sınır koordinatlarını al
            if polygon.exterior:
                poly_coords = list(polygon.exterior.coords)[:-1]  # Son noktayı çıkar (ilk noktanın tekrarı)
                
                # Her noktanın polygon sınırındaki en yakın indeksini bul
                indices = []
                for point in points:
                    min_dist = float('inf')
                    min_idx = 0
                    for i, coord in enumerate(poly_coords):
                        dist = np.linalg.norm(point[:2] - np.array(coord))
                        if dist < min_dist:
                            min_dist = dist
                            min_idx = i
                    indices.append(min_idx)
                
                # İndislere göre sırala
                sorted_indices = np.argsort(indices)
                return sorted_indices
        except Exception as e:
            logging.warning(f"Nokta sıralama hatası: {e}")
        
        return np.arange(len(points))

    def __len__(self) -> int: 
        return len(self.points)
    
    def __getitem__(self, index) -> np.ndarray: 
        return self.points[index]
    
    def __iter__(self): 
        return iter(self.points)
    
    def __repr__(self) -> str:
        return f"SahteHull(n_points={len(self)}, area={self.polygon.area:.2f})"

class HullManager:
    def __init__(self, filo_ref=None, 
                 buffer_radius: float = 12.0,
                 kanal_genislik: float = 2.0,
                 tolerans_orani: float = 0.9,
                 logger: Optional[logging.Logger] = None):
        """
        HullManager sınıfı başlatıcısı
        
        Args:
            filo_ref: Filo referansı
            buffer_radius: Yasaklı nokta buffer yarıçapı
            kanal_genislik: Kanal genişliği
            tolerans_orani: Formasyon tolerans oranı (0.0-1.0)
            logger: Loglama için logger nesnesi
        """
        self.filo = filo_ref
        self.default_buffer_radius = buffer_radius
        self.kanal_genislik = kanal_genislik
        self.tolerans_orani = tolerans_orani
        
        # Loglama ayarı
        self.logger = logger or logging.getLogger(__name__)
        if not self.logger.handlers:
            handler = logging.StreamHandler()
            formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
            handler.setFormatter(formatter)
            self.logger.addHandler(handler)
            self.logger.setLevel(logging.INFO)
    
    def yeni_hull_ve_goruntule(self, 
                              base_points: List[Tuple[float, float]], 
                              yasakli_noktalar: Optional[List[Tuple[float, float]]] = None, 
                              offset: float = 0.0, 
                              buffer_radius: Optional[float] = None,
                              kesisme_mesafe_carpani: float = 2.0,
                              min_nokta_sayisi: int = 3,
                              **kwargs) -> HullResult:
        """
        Hull oluştur ve görselleştir. Yasaklı noktaların nasıl çıkarıldığını göster.
        
        Returns:
            HullResult: Hull oluşturma sonucu
        """
        print("=" * 70)
        print("HULL OLUŞTURMA SÜRECİ - DETAYLI GÖRSELLEŞTİRME")
        print("=" * 70)
        
        # 1. Önce hull'ı oluştur
        result = self.yeni_hull(
            base_points=base_points,
            yasakli_noktalar=yasakli_noktalar,
            offset=offset,
            buffer_radius=buffer_radius,
            kesisme_mesafe_carpani=kesisme_mesafe_carpani,
            min_nokta_sayisi=min_nokta_sayisi,
            **kwargs
        )
        
        # 2. Sonuçları ekrana yazdır
        self._sonuclari_yazdir(result, base_points, yasakli_noktalar)
        
        # 3. Görselleştir
        if result.is_valid:
            self._hull_goruntule(result, base_points, yasakli_noktalar, buffer_radius)
        
        return result
    
    def yeni_hull(self, 
                  base_points: List[Tuple[float, float]], 
                  yasakli_noktalar: Optional[List[Tuple[float, float]]] = None, 
                  offset: float = 0.0, 
                  buffer_radius: Optional[float] = None,
                  kesisme_mesafe_carpani: float = 2.0,
                  min_nokta_sayisi: int = 3,
                  **kwargs) -> HullResult:
        """
        Daireyi alır ve adaları kanal açarak içinden çıkarır.
        
        Args:
            base_points: Temel noktalar [(x1, y1), (x2, y2), ...]
            yasakli_noktalar: Yasaklı noktalar [(x1, y1), (x2, y2), ...]
            offset: Polygon buffer offset değeri
            buffer_radius: Yasaklı nokta buffer yarıçapı (None ise default kullanılır)
            kesisme_mesafe_carpani: Yasaklı nokta kontrolü için mesafe çarpanı
            min_nokta_sayisi: Minimum nokta sayısı
            
        Returns:
            HullResult: Hull oluşturma sonucu
        """
        # Parametreleri kontrol et
        buffer_radius = buffer_radius or self.default_buffer_radius
        
        # Boş sonuç için
        BOS = HullResult(is_valid=False, error_message="Başlangıç noktaları yetersiz")
        
        if not base_points or len(base_points) < min_nokta_sayisi:
            BOS.error_message = f"En az {min_nokta_sayisi} nokta gerekli, {len(base_points) if base_points else 0} verildi"
            self.logger.warning(BOS.error_message)
            return BOS

        try:
            # Base points'i numpy array'e çevir
            base_points_array = np.array(base_points)
            
            # Polygon oluştur
            poly = Polygon(base_points)
            if not poly.is_valid:
                self.logger.info("Polygon geçersiz, buffer(0) uygulanıyor")
                poly = poly.buffer(0)
            
            # Alan kontrolü
            if poly.area < 1e-6:
                BOS.error_message = "Polygon alanı çok küçük"
                self.logger.warning(BOS.error_message)
                return BOS
            
            # Offset uygula (genişletme/daraltma)
            if offset != 0.0:
                self.logger.debug(f"Offset uygulanıyor: {offset}")
                poly = poly.buffer(offset)
                if poly.is_empty:
                    BOS.error_message = f"Offset ({offset}) sonrası polygon boş"
                    self.logger.warning(BOS.error_message)
                    return BOS

            # Yasaklı noktaları işle
            if yasakli_noktalar and len(yasakli_noktalar) > 0:
                poly = self._yasakli_noktalari_isle(poly, yasakli_noktalar, buffer_radius, kesisme_mesafe_carpani)
                if poly is None or poly.is_empty:
                    BOS.error_message = "Yasaklı noktalar işlendikten sonra polygon boş"
                    self.logger.warning(BOS.error_message)
                    return BOS

            # Çoklu polygon kontrolü
            if isinstance(poly, MultiPolygon):
                poly = self._multipolygon_isle(poly)
                if poly is None:
                    BOS.error_message = "MultiPolygon işleme başarısız"
                    self.logger.warning(BOS.error_message)
                    return BOS

            # Son kontrol
            if poly.is_empty or not poly.is_valid:
                BOS.error_message = "Son polygon geçersiz veya boş"
                self.logger.warning(BOS.error_message)
                return BOS

            # Noktaları ve merkezi hazırla
            new_points = np.array(poly.exterior.coords)
            
            # Merkez noktasını hesapla (representative point)
            rp = poly.representative_point()
            center = (float(rp.x), float(rp.y), 0.0)
            
            # SahteHull nesnesi oluştur
            custom_hull = SahteHull(new_points, poly)
            
            # Başarılı sonuç
            result = HullResult(
                points=new_points,
                center=center,
                hull=custom_hull,
                polygon=poly,
                is_valid=True,
                error_message=None
            )
            
            self.logger.info(f"Hull başarıyla oluşturuldu: {len(new_points)} nokta, alan={poly.area:.2f}")
            return result

        except Exception as e:
            error_msg = f"Hull oluşturma hatası: {str(e)}"
            self.logger.error(error_msg, exc_info=True)
            return HullResult(is_valid=False, error_message=error_msg)

    def _yasakli_noktalari_isle(self, 
                                poly: Polygon, 
                                yasakli_noktalar: List[Tuple[float, float]], 
                                buffer_radius: float,
                                kesisme_mesafe_carpani: float) -> Optional[Polygon]:
        """
        Yasaklı noktaları polygon'dan çıkarır.
        
        Args:
            poly: Orijinal polygon
            yasakli_noktalar: Yasaklı nokta listesi
            buffer_radius: Buffer yarıçapı
            kesisme_mesafe_carpani: Kesişme kontrolü için çarpan
            
        Returns:
            Güncellenmiş polygon veya None
        """
        try:
            # Tüm yasaklı alanları topla
            yasak_alanlar = []
            yasak_nokta_sayisi = 0
            
            for p in yasakli_noktalar:
                if len(p) >= 2:
                    yasak_nokta_sayisi += 1
                    pt_obj = Point(float(p[0]), float(p[1]))
                    
                    # Sadece polygon'a belirli mesafede olan noktaları işle
                    distance_to_poly = poly.distance(pt_obj)
                    max_distance = buffer_radius * kesisme_mesafe_carpani
                    
                    if distance_to_poly < max_distance:
                        # Yasaklı alan oluştur
                        yasak_alan = pt_obj.buffer(buffer_radius)
                        
                        # Kanal oluştur (polygon'a bağlantı için)
                        p1, p2 = nearest_points(yasak_alan, poly.exterior)
                        kanal = LineString([p1, p2]).buffer(self.kanal_genislik)
                        
                        # Yasaklı alan ve kanalı birleştir
                        kesilecek_alan = unary_union([yasak_alan, kanal])
                        yasak_alanlar.append(kesilecek_alan)
            
            self.logger.debug(f"{yasak_nokta_sayisi} yasaklı noktadan {len(yasak_alanlar)}'si işlendi")
            
            # Tüm yasaklı alanları polygon'dan çıkar
            if yasak_alanlar:
                birlesik_yasak = unary_union(yasak_alanlar)
                result_poly = poly.difference(birlesik_yasak)
                
                # Sonucu kontrol et
                if result_poly.is_empty:
                    self.logger.warning("Yasaklı alanlar çıkarıldıktan sonra polygon boş")
                    return None
                    
                return result_poly
            
            return poly  # Yasaklı alan yoksa orijinal polygon'u dön
            
        except Exception as e:
            self.logger.error(f"Yasaklı nokta işleme hatası: {e}", exc_info=True)
            return poly  # Hata durumunda orijinal polygon'u koru

    def _multipolygon_isle(self, multipoly: MultiPolygon) -> Optional[Polygon]:
        """
        MultiPolygon'ı işler ve en uygun polygon'ı seçer.
        
        Args:
            multipoly: MultiPolygon nesnesi
            
        Returns:
            Seçilen Polygon veya None
        """
        if not multipoly.geoms:
            return None
        
        # Birden fazla polygon varsa, strateji seç
        if len(multipoly.geoms) == 1:
            # Sadece bir polygon varsa onu dön
            return multipoly.geoms[0]
        else:
            # 1. En büyük alanlı polygon (varsayılan)
            en_buyuk = max(multipoly.geoms, key=lambda a: a.area)
            
            # 2. En kompakt olanı (en küçük çevre/alan oranı)
            # en_kompakt = min(multipoly.geoms, key=lambda a: a.length / a.area if a.area > 0 else float('inf'))
            
            # 3. Tüm polygon'ları birleştir (eğer birbirine yakınsa)
            # birlesik = unary_union(multipoly.geoms)
            # if isinstance(birlesik, Polygon):
            #     return birlesik
            
            self.logger.info(f"MultiPolygon: {len(multipoly.geoms)} parça, en büyük alan seçildi")
            return en_buyuk

    def _sonuclari_yazdir(self, result: HullResult, 
                         base_points: List[Tuple[float, float]], 
                         yasakli_noktalar: Optional[List[Tuple[float, float]]]):
        """Hull sonuçlarını detaylı şekilde yazdır"""
        
        print(f"\n{'═'*30} GİRDİ BİLGİLERİ {'═'*30}")
        print(f"Temel Nokta Sayısı: {len(base_points)}")
        print("Temel Noktalar:")
        for i, point in enumerate(base_points):
            print(f"  {i+1}. ({point[0]:.2f}, {point[1]:.2f})")
        
        if yasakli_noktalar:
            print(f"\nYasaklı Nokta Sayısı: {len(yasakli_noktalar)}")
            print("Yasaklı Noktalar:")
            for i, point in enumerate(yasakli_noktalar):
                print(f"  {i+1}. ({point[0]:.2f}, {point[1]:.2f})")
        else:
            print("\nYasaklı Nokta: Yok")
        
        print(f"\n{'═'*30} ÇIKTI BİLGİLERİ {'═'*30}")
        
        if result.is_valid:
            print(f"✓ HULL BAŞARIYLA OLUŞTURULDU")
            print(f"\nOluşan Hull Noktaları ({len(result.points)} adet):")
            for i, point in enumerate(result.points):
                print(f"  {i+1}. ({point[0]:.2f}, {point[1]:.2f})")
            
            print(f"\nMerkez Noktası:")
            print(f"  ({result.center[0]:.2f}, {result.center[1]:.2f}, {result.center[2]:.2f})")
            
            print(f"\nHull Özellikleri:")
            print(f"  Alan: {result.polygon.area:.2f} birim²")
            print(f"  Çevre: {result.polygon.length:.2f} birim")
            print(f"  Geçerli: {result.polygon.is_valid}")
            
            # Bounding box bilgisi
            bounds = result.polygon.bounds
            print(f"\nSınırlayıcı Kutu (Bounding Box):")
            print(f"  Min X: {bounds[0]:.2f}, Max X: {bounds[2]:.2f}")
            print(f"  Min Y: {bounds[1]:.2f}, Max Y: {bounds[3]:.2f}")
            print(f"  Genişlik: {bounds[2]-bounds[0]:.2f}, Yükseklik: {bounds[3]-bounds[1]:.2f}")
            
        else:
            print(f"✗ HULL OLUŞTURULAMADI")
            print(f"Hata: {result.error_message}")

    def _hull_goruntule(self, result: HullResult, 
                       base_points: List[Tuple[float, float]], 
                       yasakli_noktalar: Optional[List[Tuple[float, float]]],
                       buffer_radius: float):
        """Hull'ı ve yasaklı noktaların çıkarılma sürecini görselleştir"""
        
        fig, axes = plt.subplots(1, 2, figsize=(16, 8))
        fig.suptitle('Hull Oluşturma Süreci - Yasaklı Noktaların Çıkarılması', 
                    fontsize=16, fontweight='bold')
        
        # 1. SOL PANEL: Orjinal durum ve yasaklı noktalar
        ax1 = axes[0]
        ax1.set_title('1. Orjinal Polygon ve Yasaklı Noktalar', fontsize=12, fontweight='bold')
        
        # Orjinal polygon'u çiz
        original_poly = Polygon(base_points)
        if original_poly.is_valid:
            x, y = original_poly.exterior.xy
            ax1.fill(x, y, alpha=0.3, color='lightblue', label='Orjinal Polygon')
            ax1.plot(x, y, 'b-', linewidth=2, label='Polygon Kenarı')
        
        # Temel noktaları göster
        base_points_array = np.array(base_points)
        ax1.scatter(base_points_array[:, 0], base_points_array[:, 1], 
                   color='blue', s=100, zorder=5, label='Temel Noktalar')
        
        # Yasaklı noktaları göster
        if yasakli_noktalar:
            yasakli_array = np.array(yasakli_noktalar)
            ax1.scatter(yasakli_array[:, 0], yasakli_array[:, 1], 
                       color='red', s=150, marker='x', linewidth=3, 
                       zorder=6, label='Yasaklı Noktalar')
            
            # Yasaklı alanları çember olarak göster
            for point in yasakli_noktalar:
                circle = plt.Circle((point[0], point[1]), buffer_radius, 
                                   color='red', alpha=0.2, linewidth=2, 
                                   linestyle='--', fill=True)
                ax1.add_patch(circle)
                ax1.text(point[0], point[1] + buffer_radius + 1, 
                        f'R={buffer_radius}', ha='center', color='red')
        
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='upper right')
        ax1.set_aspect('equal', adjustable='box')
        ax1.set_xlabel('X Koordinatı')
        ax1.set_ylabel('Y Koordinatı')
        
        # 2. SAĞ PANEL: Oluşan hull
        ax2 = axes[1]
        ax2.set_title('2. Oluşan Hull (Yasaklı Noktalar Çıkarılmış)', 
                     fontsize=12, fontweight='bold')
        
        # Oluşan hull'ı çiz
        if result.polygon:
            x, y = result.polygon.exterior.xy
            ax2.fill(x, y, alpha=0.5, color='lightgreen', label='Oluşan Hull')
            ax2.plot(x, y, 'g-', linewidth=3, label='Hull Kenarı')
            
            # Hull noktalarını göster
            ax2.scatter(result.points[:, 0], result.points[:, 1], 
                       color='green', s=80, zorder=5, label='Hull Noktaları')
            
            # Merkez noktasını göster
            ax2.scatter(result.center[0], result.center[1], 
                       color='orange', s=200, marker='*', 
                       zorder=6, label=f'Merkez ({result.center[0]:.1f}, {result.center[1]:.1f})')
            
            # Kanal açılan bölgeleri göster (varsa)
            if yasakli_noktalar:
                for point in yasakli_noktalar:
                    pt_obj = Point(point[0], point[1])
                    distance_to_poly = result.polygon.distance(pt_obj)
                    
                    if distance_to_poly < buffer_radius * 2:  # Yakınsa
                        # Kanal bölgesini şematik olarak göster
                        ax2.scatter(point[0], point[1], 
                                   color='red', s=100, marker='x', 
                                   linewidth=2, zorder=6, 
                                   label='Çıkarılan Yasaklı Nokta')
        
        # Orjinal noktaları hafifçe göster
        ax2.scatter(base_points_array[:, 0], base_points_array[:, 1], 
                   color='blue', s=30, alpha=0.3, label='Orjinal Noktalar (Hafif)')
        
        # Bilgi metni
        info_text = f"""
        Hull Bilgisi:
        • Nokta Sayısı: {len(result.points)}
        • Alan: {result.polygon.area:.1f} birim²
        • Çevre: {result.polygon.length:.1f} birim
        • Yasaklı Nokta: {len(yasakli_noktalar) if yasakli_noktalar else 0} adet
        """
        
        ax2.text(0.02, 0.98, info_text, transform=ax2.transAxes,
                fontsize=10, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
        
        ax2.grid(True, alpha=0.3)
        ax2.legend(loc='upper right')
        ax2.set_aspect('equal', adjustable='box')
        ax2.set_xlabel('X Koordinatı')
        ax2.set_ylabel('Y Koordinatı')
        
        plt.tight_layout()
        plt.show()
        
        # 3. EK DETAY GÖRSELLER
        if yasakli_noktalar and len(yasakli_noktalar) > 0:
            self._kanal_detay_goruntule(result, yasakli_noktalar, buffer_radius)

    def _kanal_detay_goruntule(self, result: HullResult, 
                              yasakli_noktalar: List[Tuple[float, float]], 
                              buffer_radius: float):
        """Kanal açma işleminin detayını göster"""
        
        fig, ax = plt.subplots(figsize=(10, 8))
        ax.set_title('Kanal Açma İşlemi - Detay', fontsize=14, fontweight='bold')
        
        # Hull'ı çiz
        if result.polygon:
            x, y = result.polygon.exterior.xy
            ax.fill(x, y, alpha=0.3, color='lightgray', label='Hull')
            ax.plot(x, y, 'k-', linewidth=2, label='Hull Sınırı')
        
        # Her yasaklı nokta için kanal işlemini göster
        colors = ['red', 'blue', 'green', 'purple', 'orange']
        
        for idx, point in enumerate(yasakli_noktalar):
            color = colors[idx % len(colors)]
            pt_obj = Point(point[0], point[1])
            
            # Yasaklı noktayı göster
            ax.scatter(point[0], point[1], color=color, s=200, 
                      marker='x', linewidth=3, zorder=10,
                      label=f'Yasaklı Nokta {idx+1}')
            
            # Yasaklı alan çemberi
            circle = plt.Circle((point[0], point[1]), buffer_radius, 
                              color=color, alpha=0.1, linewidth=2, 
                              linestyle='--', fill=True)
            ax.add_patch(circle)
            
            # Kanal çizgisi (şematik)
            if result.polygon:
                # Hull'a en yakın noktayı bul
                p1, p2 = nearest_points(pt_obj.buffer(buffer_radius), result.polygon.exterior)
                
                # Kanal çizgisini göster
                ax.plot([p1.x, p2.x], [p1.y, p2.y], 
                       color=color, linewidth=3, linestyle=':', 
                       alpha=0.7, label=f'Kanal {idx+1}')
                
                # Kanal giriş noktaları
                ax.scatter([p1.x, p2.x], [p1.y, p2.y], 
                          color=color, s=100, marker='o', zorder=10)
                
                # Mesafe bilgisi
                distance = np.sqrt((p2.x - p1.x)**2 + (p2.y - p1.y)**2)
                mid_x = (p1.x + p2.x) / 2
                mid_y = (p1.y + p2.y) / 2
                
                ax.text(mid_x, mid_y, f'{distance:.1f}', 
                       ha='center', va='center', 
                       bbox=dict(boxstyle='round', facecolor='white', alpha=0.8),
                       fontsize=9)
        
        # Bilgi kutusu
        info_text = f"""
        Kanal Açma Süreci:
        1. Her yasaklı noktaya {buffer_radius} birim yarıçapında çember çizilir
        2. Çemberden hull sınırına en kısa hat çekilir
        3. Bu hat genişletilerek (kanal genişliği={self.kanal_genislik}) yasaklı alan hull'dan çıkarılır
        4. Yasaklı alan ve kanal, hull alanından kesilir
        """
        
        ax.text(0.02, 0.02, info_text, transform=ax.transAxes,
                fontsize=10, verticalalignment='bottom',
                bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))
        
        ax.grid(True, alpha=0.3)
        ax.legend(loc='upper right', fontsize=9)
        ax.set_aspect('equal', adjustable='box')
        ax.set_xlabel('X Koordinatı')
        ax.set_ylabel('Y Koordinatı')
        
        plt.tight_layout()
        plt.show()

    # Diğer metodlar aynı kalacak...
    def formasyon_gecerli_mi(self, 
                            test_points: List[Tuple[float, float, float]], 
                            hull: Any, 
                            formasyon_aralik: float,
                            tolerans_orani: Optional[float] = None) -> Dict[str, Any]:
        """Formasyon kontrolü (önceki versiyonla aynı)"""
        # ... önceki kod buraya gelecek
        pass
    
    def is_point_inside_hull(self, 
                             point: Tuple[float, float, float], 
                             hull: Any) -> Dict[str, Any]:
        """Nokta kontrolü (önceki versiyonla aynı)"""
        # ... önceki kod buraya gelecek
        pass

    def hull_bilgisi(self, hull: Any) -> Dict[str, Any]:
        """Hull bilgisi (önceki versiyonla aynı)"""
        # ... önceki kod buraya gelecek
        pass


# ÖRNEK KULLANIM
if __name__ == "__main__":
    print("Hull Oluşturma ve Görselleştirme Örneği")
    print("=" * 50)
    
    # HullManager oluştur
    manager = HullManager(
        buffer_radius=8.0,
        kanal_genislik=1.5,
        tolerans_orani=0.85,
        logger=None
    )
    
    # TEST 1: Basit kare hull
    print("\nTEST 1: Basit Kare Hull")
    print("-" * 30)
    
    base_points_1 = [(0, 0), (20, 0), (20, 20), (0, 20)]
    yasakli_noktalar_1 = [(5, 5), (15, 5), (10, 15), (18, 18)]
    
    result1 = manager.yeni_hull_ve_goruntule(
        base_points=base_points_1,
        yasakli_noktalar=yasakli_noktalar_1,
        offset=0.5,
        buffer_radius=6.0
    )
    
    # TEST 2: Daha karmaşık şekil
    print("\n\nTEST 2: Karmaşık Şekil")
    print("-" * 30)
    
    base_points_2 = [
        (0, 0), (25, 0), (25, 15), (20, 20), 
        (15, 25), (10, 20), (5, 15), (0, 10)
    ]
    yasakli_noktalar_2 = [(8, 8), (12, 12), (20, 10), (5, 18)]
    
    result2 = manager.yeni_hull_ve_goruntule(
        base_points=base_points_2,
        yasakli_noktalar=yasakli_noktalar_2,
        offset=1.0,
        buffer_radius=5.0
    )
    
    # TEST 3: Yasaklı nokta olmadan
    print("\n\nTEST 3: Yasaklı Nokta Olmadan")
    print("-" * 30)
    
    base_points_3 = [(0, 0), (15, 0), (15, 15), (0, 15)]
    
    result3 = manager.yeni_hull_ve_goruntule(
        base_points=base_points_3,
        yasakli_noktalar=None,  # Yasaklı nokta yok
        offset=0.0,
        buffer_radius=5.0
    )
    
    print("\n" + "="*50)
    print("TÜM TESTLER TAMAMLANDI")
    print("="*50)
```

```python
import numpy as np
import math
import logging
from shapely.geometry import Point, Polygon, MultiPolygon, LineString, MultiPoint, box
from shapely.ops import unary_union, nearest_points
from shapely.prepared import prep
from scipy.spatial.distance import pdist
from typing import List, Tuple, Optional, Union, Dict, Any
from dataclasses import dataclass

# Scipy Kontrolü
try:
    from scipy.spatial import ConvexHull
    SCIPY_AVAILABLE = True
except ImportError:
    SCIPY_AVAILABLE = False

@dataclass
class HullResult:
    """Hull oluşturma sonucunu saklamak için veri sınıfı"""
    points: Optional[np.ndarray] = None  # 100 noktalı çevre
    center: Optional[Tuple[float, float, float]] = None
    hull: Any = None
    polygon: Optional[Polygon] = None
    is_valid: bool = False
    error_message: Optional[str] = None

class SahteHull:
    """
    Hem bir liste gibi davranan hem de Shapely Polygon taşıyan 
    ve len() destekleyen Wrapper sınıfı.
    """
    def __init__(self, points: np.ndarray, polygon_obj: Polygon):
        self.points = points  # 100 noktalı çevre
        self.polygon = polygon_obj
        # Noktaları polygon sınırına göre doğru sırala
        self.vertices = np.arange(len(points))  # 0-99
        # Polygon'ın köşeleri için simplices oluştur
        n = len(points)
        self.simplices = np.array([[i, (i + 1) % n] for i in range(n)])

    def __len__(self) -> int: 
        return len(self.points)
    
    def __getitem__(self, index) -> np.ndarray: 
        return self.points[index]
    
    def __iter__(self): 
        return iter(self.points)
    
    def __repr__(self) -> str:
        return f"SahteHull(n_points={len(self)}, area={self.polygon.area:.2f})"

class HullManager:
    def __init__(self, filo_ref=None, 
                 buffer_radius: float = 12.0,
                 kanal_genislik: float = 2.0,
                 tolerans_orani: float = 0.9,
                 logger: Optional[logging.Logger] = None):
        """
        HullManager sınıfı başlatıcısı
        
        Args:
            filo_ref: Filo referansı
            buffer_radius: Yasaklı nokta buffer yarıçapı
            kanal_genislik: Kanal genişliği
            tolerans_orani: Formasyon tolerans oranı (0.0-1.0)
            logger: Loglama için logger nesnesi
        """
        self.filo = filo_ref
        self.default_buffer_radius = buffer_radius
        self.kanal_genislik = kanal_genislik
        self.tolerans_orani = tolerans_orani
        
        # Loglama ayarı
        self.logger = logger or logging.getLogger(__name__)
        if not self.logger.handlers:
            handler = logging.StreamHandler()
            formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
            handler.setFormatter(formatter)
            self.logger.addHandler(handler)
            self.logger.setLevel(logging.INFO)
    
    def hull_olustur(self, 
                    base_points: List[Tuple[float, float]], 
                    yasakli_noktalar: Optional[List[Tuple[float, float]]] = None, 
                    offset: float = 0.0, 
                    buffer_radius: Optional[float] = None,
                    kesisme_mesafe_carpani: float = 2.0,
                    nokta_sayisi: int = 100,
                    min_nokta_sayisi: int = 3,
                    **kwargs) -> HullResult:
        """
        Ana hull oluşturma fonksiyonu - Her zaman nokta_sayisi kadar nokta döndürür.
        
        Args:
            base_points: Temel noktalar [(x1, y1), (x2, y2), ...]
            yasakli_noktalar: Yasaklı noktalar [(x1, y1), (x2, y2), ...]
            offset: Polygon buffer offset değeri
            buffer_radius: Yasaklı nokta buffer yarıçapı
            kesisme_mesafe_carpani: Yasaklı nokta kontrolü için mesafe çarpanı
            nokta_sayisi: Döndürülecek nokta sayısı (varsayılan: 100)
            min_nokta_sayisi: Minimum nokta sayısı
            
        Returns:
            HullResult: 100 noktalı hull sonucu
        """
        print("=" * 60)
        print("100 NOKTA HULL OLUŞTURMA")
        print("=" * 60)
        
        # Parametreleri kontrol et
        buffer_radius = buffer_radius or self.default_buffer_radius
        
        # Boş sonuç için
        BOS = HullResult(is_valid=False, error_message="Başlangıç noktaları yetersiz")
        
        if not base_points or len(base_points) < min_nokta_sayisi:
            BOS.error_message = f"En az {min_nokta_sayisi} nokta gerekli, {len(base_points) if base_points else 0} verildi"
            self.logger.warning(BOS.error_message)
            return BOS

        try:
            # 1. Temel polygon'u oluştur
            poly = self._temel_polygon_olustur(base_points, offset)
            if poly is None:
                BOS.error_message = "Temel polygon oluşturulamadı"
                return BOS
            
            # 2. Yasaklı noktaları işle
            if yasakli_noktalar and len(yasakli_noktalar) > 0:
                poly = self._yasakli_noktalari_isle(poly, yasakli_noktalar, buffer_radius, kesisme_mesafe_carpani)
                if poly is None or poly.is_empty:
                    BOS.error_message = "Yasaklı noktalar işlendikten sonra polygon boş"
                    return BOS

            # 3. Çoklu polygon kontrolü
            if isinstance(poly, MultiPolygon):
                poly = self._multipolygon_isle(poly)
                if poly is None:
                    BOS.error_message = "MultiPolygon işleme başarısız"
                    return BOS

            # 4. Son kontroller
            if poly.is_empty or not poly.is_valid:
                BOS.error_message = "Son polygon geçersiz veya boş"
                return BOS

            # 5. 100 noktalı çevre oluştur
            noktalar_100 = self._100_noktali_cevre_olustur(poly, nokta_sayisi)
            if noktalar_100 is None:
                BOS.error_message = "100 noktalı çevre oluşturulamadı"
                return BOS

            # 6. Merkez noktasını hesapla
            center = self._merkez_noktasi_hesapla(poly)
            
            # 7. SahteHull nesnesi oluştur
            custom_hull = SahteHull(noktalar_100, poly)
            
            # 8. Sonuçları hazırla
            result = HullResult(
                points=noktalar_100,
                center=center,
                hull=custom_hull,
                polygon=poly,
                is_valid=True,
                error_message=None
            )
            
            # 9. Sonuçları yazdır
            self._sonuclari_yazdir(result, base_points, yasakli_noktalar, buffer_radius)
            
            self.logger.info(f"Hull başarıyla oluşturuldu: {nokta_sayisi} nokta, alan={poly.area:.2f}")
            return result

        except Exception as e:
            error_msg = f"Hull oluşturma hatası: {str(e)}"
            self.logger.error(error_msg, exc_info=True)
            return HullResult(is_valid=False, error_message=error_msg)
    
    def _temel_polygon_olustur(self, base_points: List[Tuple[float, float]], offset: float) -> Optional[Polygon]:
        """Temel polygon'u oluşturur"""
        try:
            poly = Polygon(base_points)
            if not poly.is_valid:
                self.logger.info("Polygon geçersiz, buffer(0) uygulanıyor")
                poly = poly.buffer(0)
            
            # Alan kontrolü
            if poly.area < 1e-6:
                self.logger.warning("Polygon alanı çok küçük")
                return None
            
            # Offset uygula
            if offset != 0.0:
                self.logger.debug(f"Offset uygulanıyor: {offset}")
                poly = poly.buffer(offset)
                if poly.is_empty:
                    self.logger.warning(f"Offset ({offset}) sonrası polygon boş")
                    return None
            
            return poly
            
        except Exception as e:
            self.logger.error(f"Temel polygon oluşturma hatası: {e}")
            return None
    
    def _yasakli_noktalari_isle(self, 
                                poly: Polygon, 
                                yasakli_noktalar: List[Tuple[float, float]], 
                                buffer_radius: float,
                                kesisme_mesafe_carpani: float) -> Optional[Polygon]:
        """
        Yasaklı noktaları polygon'dan çıkarır.
        """
        try:
            yasak_alanlar = []
            yasak_nokta_sayisi = 0
            islenen_yasakli_noktalar = []
            
            for p in yasakli_noktalar:
                if len(p) >= 2:
                    yasak_nokta_sayisi += 1
                    pt_obj = Point(float(p[0]), float(p[1]))
                    
                    # Sadece polygon'a belirli mesafede olan noktaları işle
                    distance_to_poly = poly.distance(pt_obj)
                    max_distance = buffer_radius * kesisme_mesafe_carpani
                    
                    if distance_to_poly < max_distance:
                        islenen_yasakli_noktalar.append(p)
                        
                        # Yasaklı alan oluştur
                        yasak_alan = pt_obj.buffer(buffer_radius)
                        
                        # Kanal oluştur (polygon'a bağlantı için)
                        p1, p2 = nearest_points(yasak_alan, poly.exterior)
                        kanal = LineString([p1, p2]).buffer(self.kanal_genislik)
                        
                        # Yasaklı alan ve kanalı birleştir
                        kesilecek_alan = unary_union([yasak_alan, kanal])
                        yasak_alanlar.append(kesilecek_alan)
            
            self.logger.debug(f"{yasak_nokta_sayisi} yasaklı noktadan {len(yasak_alanlar)}'si işlendi")
            
            # Tüm yasaklı alanları polygon'dan çıkar
            if yasak_alanlar:
                birlesik_yasak = unary_union(yasak_alanlar)
                result_poly = poly.difference(birlesik_yasak)
                
                if result_poly.is_empty:
                    self.logger.warning("Yasaklı alanlar çıkarıldıktan sonra polygon boş")
                    return None
                    
                return result_poly
            
            return poly  # Yasaklı alan yoksa orijinal polygon'u dön
            
        except Exception as e:
            self.logger.error(f"Yasaklı nokta işleme hatası: {e}")
            return poly
    
    def _multipolygon_isle(self, multipoly: MultiPolygon) -> Optional[Polygon]:
        """MultiPolygon'ı işler ve en uygun polygon'ı seçer."""
        if not multipoly.geoms:
            return None
        
        if len(multipoly.geoms) == 1:
            return multipoly.geoms[0]
        else:
            # En büyük alanlı polygon'ı seç
            en_buyuk = max(multipoly.geoms, key=lambda a: a.area)
            self.logger.info(f"MultiPolygon: {len(multipoly.geoms)} parça, en büyük alan seçildi")
            return en_buyuk
    
    def _100_noktali_cevre_olustur(self, polygon: Polygon, nokta_sayisi: int = 100) -> Optional[np.ndarray]:
        """
        Polygon çevresini eşit aralıklı nokta_sayisi kadar nokta ile temsil eder.
        
        Args:
            polygon: Shapely Polygon nesnesi
            nokta_sayisi: Oluşturulacak nokta sayısı
            
        Returns:
            (nokta_sayisi, 2) boyutunda numpy array
        """
        try:
            # Polygon çevresini al
            if isinstance(polygon, Polygon):
                # Çevre uzunluğunu hesapla
                cevre_uzunlugu = polygon.length
                
                # Eğer polygon çok küçükse
                if cevre_uzunlugu < 1e-6:
                    self.logger.warning("Polygon çevresi çok küçük")
                    return None
                
                # Eşit aralıklı noktalar oluştur
                noktalar = []
                
                # Çevre boyunca ilerle
                for i in range(nokta_sayisi):
                    # Çevrenin yüzdesi kadar ilerle
                    mesafe_orani = i / nokta_sayisi
                    
                    # Çevre boyunca mesafe oranına karşılık gelen noktayı bul
                    nokta = polygon.exterior.interpolate(mesafe_orani * cevre_uzunlugu)
                    noktalar.append([nokta.x, nokta.y])
                
                return np.array(noktalar)
            
            return None
            
        except Exception as e:
            self.logger.error(f"100 noktalı çevre oluşturma hatası: {e}")
            return None
    
    def _merkez_noktasi_hesapla(self, polygon: Polygon) -> Tuple[float, float, float]:
        """Polygon'un merkez noktasını hesaplar"""
        try:
            centroid = polygon.centroid
            return (float(centroid.x), float(centroid.y), 0.0)
        except:
            # Fallback: representative point
            rp = polygon.representative_point()
            return (float(rp.x), float(rp.y), 0.0)
    
    def _sonuclari_yazdir(self, result: HullResult, 
                         base_points: List[Tuple[float, float]], 
                         yasakli_noktalar: Optional[List[Tuple[float, float]]],
                         buffer_radius: float):
        """Sonuçları detaylı şekilde yazdır"""
        
        print(f"\n{'═'*30} GİRDİ PARAMETRELERİ {'═'*30}")
        print(f"• Temel Nokta Sayısı: {len(base_points)}")
        print(f"• Yasaklı Nokta Sayısı: {len(yasakli_noktalar) if yasakli_noktalar else 0}")
        print(f"• Buffer Radius: {buffer_radius:.1f}")
        print(f"• Kanal Genişliği: {self.kanal_genislik:.1f}")
        
        if result.is_valid:
            print(f"\n{'═'*30} HULL SONUÇLARI {'═'*30}")
            print(f"✓ BAŞARILI - {len(result.points)} NOKTA")
            
            # İlk 5 ve son 5 noktayı göster
            print(f"\nHull Noktaları (İlk 5 ve Son 5):")
            for i in range(5):
                print(f"  {i+1:3d}. ({result.points[i, 0]:8.3f}, {result.points[i, 1]:8.3f})")
            print("  ...")
            for i in range(-5, 0):
                idx = len(result.points) + i
                print(f"  {idx+1:3d}. ({result.points[i, 0]:8.3f}, {result.points[i, 1]:8.3f})")
            
            print(f"\nMerkez Noktası:")
            print(f"  ({result.center[0]:.3f}, {result.center[1]:.3f}, {result.center[2]:.3f})")
            
            print(f"\nGeometrik Özellikler:")
            print(f"  Alan: {result.polygon.area:.2f} birim²")
            print(f"  Çevre: {result.polygon.length:.2f} birim")
            print(f"  Nokta Yoğunluğu: {len(result.points)/result.polygon.length:.3f} nokta/birim")
            
            # İstatistikler
            if len(result.points) > 1:
                mesafeler = np.linalg.norm(np.diff(result.points, axis=0), axis=1)
                print(f"\nNokta Mesafe İstatistikleri:")
                print(f"  Ortalama Mesafe: {np.mean(mesafeler):.3f} birim")
                print(f"  Minimum Mesafe: {np.min(mesafeler):.3f} birim")
                print(f"  Maksimum Mesafe: {np.max(mesafeler):.3f} birim")
            
            # Yasaklı nokta durumu
            if yasakli_noktalar and result.polygon:
                print(f"\nYasaklı Nokta Analizi:")
                yasakli_durumlar = []
                for i, point in enumerate(yasakli_noktalar):
                    pt = Point(point[0], point[1])
                    mesafe = result.polygon.distance(pt)
                    durum = "ÇIKARILDI" if mesafe < buffer_radius else "DIŞARIDA"
                    yasakli_durumlar.append((i+1, point, mesafe, durum))
                
                for idx, point, mesafe, durum in yasakli_durumlar:
                    print(f"  {idx}. ({point[0]:.1f}, {point[1]:.1f}): {durum} (mesafe: {mesafe:.2f})")
        else:
            print(f"\n✗ HATA: {result.error_message}")

    def formasyon_kontrol(self, 
                         test_points: List[Tuple[float, float, float]], 
                         hull_result: HullResult, 
                         min_mesafe: float = 3.0) -> Dict[str, Any]:
        """
        Formasyon noktalarının hull içinde olup olmadığını kontrol eder.
        
        Args:
            test_points: Test edilecek noktalar [(x, y, z), ...]
            hull_result: Hull oluşturma sonucu
            min_mesafe: Minimum mesafe kontrolü
            
        Returns:
            Kontrol sonucu sözlüğü
        """
        sonuc = {
            'gecerli': False,
            'sebep': '',
            'icinde_olanlar': [],
            'disarida_olanlar': [],
            'mesafe_uyari': []
        }
        
        if not hull_result.is_valid:
            sonuc['sebep'] = 'Hull geçersiz'
            return sonuc
        
        if not test_points:
            sonuc['sebep'] = 'Test noktası yok'
            return sonuc
        
        try:
            # Her noktayı kontrol et
            for i, point in enumerate(test_points):
                nokta_2d = Point(point[0], point[1])
                if hull_result.polygon.contains(nokta_2d):
                    sonuc['icinde_olanlar'].append(i)
                else:
                    sonuc['disarida_olanlar'].append(i)
            
            # Mesafe kontrolü
            if len(test_points) > 1:
                noktalar_2d = np.array([(p[0], p[1]) for p in test_points])
                for i in range(len(noktalar_2d)):
                    for j in range(i+1, len(noktalar_2d)):
                        mesafe = np.linalg.norm(noktalar_2d[i] - noktalar_2d[j])
                        if mesafe < min_mesafe:
                            sonuc['mesafe_uyari'].append({
                                'nokta1': i,
                                'nokta2': j,
                                'mesafe': mesafe
                            })
            
            # Sonuç değerlendirme
            if len(sonuc['disarida_olanlar']) == 0:
                if len(sonuc['mesafe_uyari']) == 0:
                    sonuc['gecerli'] = True
                    sonuc['sebep'] = 'Tüm noktalar hull içinde ve mesafeler uygun'
                else:
                    sonuc['gecerli'] = False
                    sonuc['sebep'] = f"{len(sonuc['mesafe_uyari'])} çift için mesafe çok yakın"
            else:
                sonuc['gecerli'] = False
                sonuc['sebep'] = f"{len(sonuc['disarida_olanlar'])} nokta hull dışında"
            
            return sonuc
            
        except Exception as e:
            sonuc['sebep'] = f"Kontrol hatası: {str(e)}"
            return sonuc

    def hull_noktalarini_al(self, hull_result: HullResult) -> np.ndarray:
        """Hull noktalarını döndürür"""
        if hull_result.is_valid:
            return hull_result.points
        return np.array([])

    def hull_merkezini_al(self, hull_result: HullResult) -> Tuple[float, float, float]:
        """Hull merkezini döndürür"""
        if hull_result.is_valid:
            return hull_result.center
        return (0.0, 0.0, 0.0)

    def hull_bilgilerini_al(self, hull_result: HullResult) -> Dict[str, Any]:
        """Hull hakkında detaylı bilgi döndürür"""
        if not hull_result.is_valid:
            return {'hata': hull_result.error_message}
        
        return {
            'nokta_sayisi': len(hull_result.points),
            'merkez': hull_result.center,
            'alan': hull_result.polygon.area,
            'cevre': hull_result.polygon.length,
            'gecerli': True,
            'bounds': hull_result.polygon.bounds
        }


# KULLANIM ÖRNEĞİ
if __name__ == "__main__":
    print("100 NOKTA HULL OLUŞTURUCU - TEST")
    print("=" * 50)
    
    # HullManager oluştur
    manager = HullManager(
        buffer_radius=10.0,
        kanal_genislik=1.5,
        tolerans_orani=0.85
    )
    
    # TEST 1: Kare hull
    print("\nTEST 1: Kare Hull")
    
    base_points = [(0, 0), (30, 0), (30, 30), (0, 30)]
    yasakli_noktalar = [(5, 5), (25, 5), (15, 25), (28, 28)]
    
    result = manager.hull_olustur(
        base_points=base_points,
        yasakli_noktalar=yasakli_noktalar,
        offset=1.0,
        buffer_radius=8.0,
        nokta_sayisi=100
    )
    
    # Sonuçları kullan
    if result.is_valid:
        print(f"\n✓ Hull başarıyla oluşturuldu!")
        
        # Noktaları al
        noktalar = manager.hull_noktalarini_al(result)
        print(f"Alınan nokta sayısı: {len(noktalar)}")
        print(f"İlk 3 nokta: {noktalar[:3]}")
        print(f"Son 3 nokta: {noktalar[-3:]}")
        
        # Merkezi al
        merkez = manager.hull_merkezini_al(result)
        print(f"Merkez: {merkez}")
        
        # Bilgileri al
        bilgiler = manager.hull_bilgilerini_al(result)
        print(f"Hull Alanı: {bilgiler['alan']:.2f}")
        print(f"Hull Çevresi: {bilgiler['cevre']:.2f}")
        
        # Formasyon kontrolü
        test_noktalar = [
            (2, 2, 0),
            (8, 8, 0),
            (20, 20, 0),
            (25, 25, 0)
        ]
        
        kontrol = manager.formasyon_kontrol(test_noktalar, result, min_mesafe=5.0)
        print(f"\nFormasyon Kontrolü:")
        print(f"  Geçerli: {kontrol['gecerli']}")
        print(f"  Sebep: {kontrol['sebep']}")
        print(f"  İçinde olanlar: {kontrol['icinde_olanlar']}")
        print(f"  Dışarıda olanlar: {kontrol['disarida_olanlar']}")
    
    else:
        print(f"\n✗ Hata: {result.error_message}")
    
    # TEST 2: Daha karmaşık şekil
    print("\n\nTEST 2: Karmaşık Şekil")
    
    base_points2 = [
        (0, 0), (40, 0), (40, 20), (35, 25),
        (30, 30), (20, 35), (10, 30), (5, 25), (0, 20)
    ]
    
    yasakli_noktalar2 = [(10, 10), (30, 10), (20, 20), (35, 15)]
    
    result2 = manager.hull_olustur(
        base_points=base_points2,
        yasakli_noktalar=yasakli_noktalar2,
        offset=0.5,
        buffer_radius=6.0,
        nokta_sayisi=100
    )
    
    if result2.is_valid:
        print(f"\n✓ İkinci hull başarıyla oluşturuldu!")
        print(f"Nokta sayısı: {len(result2.points)}")
        print(f"Alan: {result2.polygon.area:.2f}")
```
